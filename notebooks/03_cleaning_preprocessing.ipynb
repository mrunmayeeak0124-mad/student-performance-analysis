```python
import pandas as pd
```

```python
df = pd.read_csv("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/data/raw/student-mat.csv", sep=';')
df.isnull().sum()        # will show 0 everywhere — expected, UCI data is clean
df.duplicated().sum()
print(df.duplicated().sum())    # check for duplicate rows
```

```python
demo = df.copy()
import numpy as np
demo.loc[demo.sample(10).index, 'absences'] = np.nan   # artificially introduce nulls
demo['absences'].fillna(demo['absences'].mean(), inplace=True)   # show mean-fill
```

```python
from scipy import stats
z_scores = stats.zscore(df['absences'])
outliers = df[abs(z_scores) > 3]
print(outliers[['absences']])
```

```python
Q1 = df['absences'].quantile(0.25)
Q3 = df['absences'].quantile(0.75)
IQR = Q3 - Q1
lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR
outliers_iqr = df[(df['absences'] < lower) | (df['absences'] > upper)]
```

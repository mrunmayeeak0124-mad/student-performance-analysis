# Student Academic Performance Prediction — Project Setup

## Objective
To analyze how study habits, lifestyle, demographic, and family/school factors 
relate to student academic performance, and to build models that predict a 
student's final grade (G3) and pass/fail status.

## Research Questions
1. Does study time have a statistically significant relationship with final grade?
2. Do past failures significantly affect final grade?
3. Is there a significant difference in grades based on parental education?
4. Can we predict G3 using demographic/lifestyle factors alone (without G1/G2)?
5. How much better is prediction when G1/G2 are included vs excluded?

## Dataset
UCI Student Performance Dataset — Mathematics subject (student-mat.csv), 395 
students, 33 attributes including grades, demographics, family background, 
and lifestyle factors.

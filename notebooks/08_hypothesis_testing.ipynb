```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

df = pd.read_csv("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/data/processed/student_clean.csv")
print("Loaded cleaned dataset. Shape:", df.shape)
```

# Research Question: Is the average final grade (G3) of our sample 
# significantly different from a hypothesized population mean of 10 
# (the passing threshold on a 20-point scale)?

# H0: mean(G3) = 10
# H1: mean(G3) != 10

```python
population_mean = 10
sample = df['G3']

t_stat, p_value = stats.ttest_1samp(sample, population_mean)

print("--- One-Sample T-Test: G3 vs Population Mean (10) ---")
print(f"Sample Mean G3: {sample.mean():.3f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.6f}")
```

```python
alpha = 0.05
if p_value < alpha:
    print("\nConclusion: Reject H0 (p < 0.05).")
    print("The sample mean G3 is significantly different from the assumed population mean of 10.")
else:
    print("\nConclusion: Fail to reject H0 (p >= 0.05).")
    print("No significant difference found between sample mean and population mean of 10.")
```

# Independent t-test: students with internet access vs without

```python
with_internet = df[df['internet'] == 'yes']['G3']
without_internet = df[df['internet'] == 'no']['G3']
t_stat2, p_value2 = stats.ttest_ind(with_internet, without_internet, equal_var=False)
print("\n--- Independent T-Test: Internet Access vs G3 ---")
print(f"Mean G3 (With Internet): {with_internet.mean():.2f}")
print(f"Mean G3 (Without Internet): {without_internet.mean():.2f}")
print(f"T-statistic: {t_stat2:.4f}, P-value: {p_value2:.6f}")
```

```python
# Test 3: One-way ANOVA, G3 across study-time groups
groups = [g['G3'].values for _, g in df.groupby('studytime')]
f_stat, p_anova = stats.f_oneway(*groups)
print("--- One-Way ANOVA: G3 across study-time groups ---")
print(df.groupby('studytime')['G3'].agg(['count', 'mean']).round(2))
print(f"F-statistic: {f_stat:.4f}, P-value: {p_anova:.6f}")
print("Decision:", "Reject H0" if p_anova < 0.05 else "Fail to reject H0")

lev_stat, lev_p = stats.levene(*groups)      # equal-variance assumption
print(f"Levene test: statistic={lev_stat:.4f}, p={lev_p:.4f}")
```

```python
# Test 4: Pearson correlation significance, absences vs G3
r, p_r = stats.pearsonr(df['absences'], df['G3'])
print("--- Pearson Correlation Test: absences vs G3 ---")
print(f"r = {r:.4f}, P-value: {p_r:.6f}")
print("Decision:", "Reject H0" if p_r < 0.05 else "Fail to reject H0")
```

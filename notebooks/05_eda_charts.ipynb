```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/data/processed/student_clean.csv")
print("Loaded cleaned dataset. Shape:", df.shape)
```

The overall spread/shape of performance

```python
plt.figure(figsize=(8,5))
sns.histplot(df['G3'], bins=15, kde=True, color='teal')
plt.title("Distribution of Final Grade (G3)")
plt.xlabel("G3"); plt.ylabel("Count")
plt.savefig("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/outputs/figures/01_g3_distribution.png")
plt.show()
print(df['G3'].describe())
```

Box plot (studytime is ordinal 1-4, categorical-like)

```python
plt.figure(figsize=(7,5))
sns.boxplot(x='studytime', y='G3', data=df, palette='Set2')
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time Category (1=<2hrs, 4=>10hrs)")
plt.ylabel("G3")
plt.savefig("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/outputs/figures/02_studytime_vs_g3.png")
plt.show()
print(df.groupby('studytime')['G3'].mean())
```

Scatter plot

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='absences', y='G3', data=df, alpha=0.6)
plt.title("Absences vs Final Grade")
plt.savefig("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/outputs/figures/03_absences_vs_g3.png")
plt.show()
print("Correlation (absences, G3):", df['absences'].corr(df['G3']))
```

Box plot

```python
plt.figure(figsize=(7,5))
sns.boxplot(x='failures', y='G3', data=df, palette='Reds')
plt.title("Past Failures vs Final Grade")
plt.savefig("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/outputs/figures/04_failures_vs_g3.png")
plt.show()
print(df.groupby('failures')['G3'].mean())
```

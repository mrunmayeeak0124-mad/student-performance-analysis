```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python

df = pd.read_csv("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/data/processed/student_clean.csv")

# Plot distribution of G3 with histogram + KDE
plt.figure(figsize=(8,5))
sns.histplot(df['G3'], bins=15, kde=True, color='steelblue')
mean_g3 = df['G3'].mean()
std_g3 = df['G3'].std()
plt.axvline(mean_g3, color='red', linestyle='--', label=f"Mean = {mean_g3:.2f}")
plt.title("Probability Distribution of Final Grade (G3)")
plt.xlabel("G3 (Final Grade)")
plt.ylabel("Frequency / Density")
plt.legend()
plt.savefig("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/outputs/figures/g3_distribution.png")
plt.show()
```

```python
print(f"Mean G3: {mean_g3:.2f}")
print(f"Std Dev G3: {std_g3:.2f}")
print(f"Skewness: {df['G3'].skew():.3f}")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/data/processed/student_clean.csv")
```

```python
numeric_cols = ['age','Medu','Fedu','traveltime','studytime','failures',
                 'famrel','freetime','goout','Dalc','Walc','health',
                 'absences','G1','G2','G3']

corr_matrix = df[numeric_cols].corr(method='pearson')
print("--- Correlation Matrix ---")
print(corr_matrix['G3'].sort_values(ascending=False))

plt.figure(figsize=(12,9))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix — Numeric Variables")
plt.savefig("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/outputs/figures/15_correlation_heatmap.png")
plt.show()

corr_matrix.to_csv("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/outputs/tables/correlation_matrix.csv")
```

```python
spearman_corr = df[numeric_cols].corr(method='spearman')
print(spearman_corr['G3'].sort_values(ascending=False))
```

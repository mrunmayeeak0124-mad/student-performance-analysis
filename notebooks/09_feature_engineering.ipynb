```python
import pandas as pd
df = pd.read_csv("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/data/processed/student_clean.csv")
```

# Pass/Fail target

```python
df['pass_fail'] = (df['G3'] >= 10).astype(int)
```

# Performance category (Low/Medium/High)

```python
def categorize(g3):
    if g3 < 10: return 'Low'
    elif g3 < 15: return 'Medium'
    else: return 'High'
df['performance_category'] = df['G3'].apply(categorize)
```

# Study time group (already ordinal, but make it readable)

```python
studytime_map = {1:'<2hrs', 2:'2-5hrs', 3:'5-10hrs', 4:'>10hrs'}
df['studytime_label'] = df['studytime'].map(studytime_map)

print(df[['G3','pass_fail','performance_category','studytime_label']].head(10))
df.to_csv("D:/Mrunmayee/OneDrive/Desktop/student-performance-project/data/processed/student_features.csv", index=False)
```
